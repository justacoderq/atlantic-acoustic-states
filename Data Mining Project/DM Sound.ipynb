```python
import glob
import os
import pandas as pd

base = "/content/drive/MyDrive/Atlantic_Ocean_Project/SB01_TOL"
files = glob.glob(base + "/**/*.csv", recursive=True)

all_times = []
skipped = []

for file in files:
    try:
        df = pd.read_csv(file, low_memory=False)
        df.columns = [str(c).strip() for c in df.columns]

        # Search for likely date/time columns
        candidates = [
            c for c in df.columns
            if any(word in c.lower() for word in
                   ["time", "date", "timestamp", "datetime"])
            and "frequency" not in c.lower()
        ]

        if not candidates:
            skipped.append((os.path.basename(file), list(df.columns)))
            continue

        time_col = candidates[0]
        parsed = pd.to_datetime(df[time_col], errors="coerce", utc=True)
        all_times.append(parsed.dropna())

    except Exception as error:
        skipped.append((os.path.basename(file), str(error)))

if all_times:
    times = pd.concat(all_times).drop_duplicates().sort_values()

    start = times.min()
    end = times.max()
    calendar_years = (end - start).total_seconds() / (365.25 * 24 * 3600)

    print("Files found:", len(files))
    print("Files containing dates:", len(all_times))
    print("Earliest observation:", start)
    print("Latest observation:", end)
    print(f"Calendar span: {calendar_years:.2f} years")
    print("Unique timestamps:", len(times))
else:
    print("No readable date columns were found.")

print("\nSkipped files and their columns:")
for item in skipped[:10]:
    print(item)
```

```python
import glob
import pandas as pd

base = "/content/drive/MyDrive/Atlantic_Ocean_Project/SB01_TOL"
files = glob.glob(base + "/**/*.csv", recursive=True)

all_times = []
deployment_summary = []

for file in files:
    df = pd.read_csv(file, low_memory=False)

    # The first column contains the timestamps
    time_col = "yyyy-mm-ddTHH:MM:SSZ"
    timestamps = pd.to_datetime(df[time_col], errors="coerce", utc=True).dropna()

    all_times.append(timestamps)

    deployment_summary.append({
        "file": file.split("/")[-1],
        "start": timestamps.min(),
        "end": timestamps.max(),
        "hourly_rows": len(timestamps)
    })

summary = pd.DataFrame(deployment_summary).sort_values("start")
display(summary)

times = pd.concat(all_times).dropna().drop_duplicates().sort_values()

start = times.min()
end = times.max()
span_years = (end - start).total_seconds() / (365.25 * 24 * 3600)
recorded_years = len(times) / (24 * 365.25)

print("\nOVERALL DATA")
print("Earliest observation:", start)
print("Latest observation:", end)
print(f"Calendar span: {span_years:.2f} years")
print("Unique hourly observations:", len(times))
print(f"Actual recorded duration: {recorded_years:.2f} years")
print("Number of deployments:", len(summary))
```
